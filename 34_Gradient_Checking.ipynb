```python
import numpy as np

# داده
x = np.array([1, 2, 3, 4])
y = np.array([2, 4, 6, 8])
w = 1.5

# تابع هزینه
def cost_function(w, x, y):
    y_pred = w * x
    return np.mean((y_pred - y) ** 2)

# گرادیان تحلیلی (دستی)
def analytical_gradient(w, x, y):
    return 2 * np.mean((w * x - y) * x)

# گرادیان عددی (Gradient Checking)
def numerical_gradient(w, x, y, epsilon=1e-7):
    cost_plus = cost_function(w + epsilon, x, y)
    cost_minus = cost_function(w - epsilon, x, y)
    return (cost_plus - cost_minus) / (2 * epsilon)

# محاسبه
grad_analytical = analytical_gradient(w, x, y)
grad_numerical = numerical_gradient(w, x, y)

print("Analytical Gradient:", grad_analytical)
print("Numerical Gradient :", grad_numerical)
print("Difference:", abs(grad_analytical - grad_numerical))
```
