```python
import numpy as np

# ----- 1) توابع پایه -----
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(z):
    s = sigmoid(z)
    return s * (1 - s)

# ----- 2) وزن‌ها -----
np.random.seed(42)
W1 = np.random.randn(3, 2) * 0.01   # (3 نورون، 2 ورودی)
b1 = np.zeros((3, 1))
W2 = np.random.randn(1, 3) * 0.01   # (1 خروجی، 3 نورون مخفی)
b2 = np.zeros((1, 1))

# ----- 3) Forward -----
def forward(X):
    z1 = W1 @ X + b1
    a1 = sigmoid(z1)

    z2 = W2 @ a1 + b2
    a2 = sigmoid(z2)

    cache = (X, z1, a1, z2, a2)  # ذخیره برای backprop
    return a2, cache

# ----- 4) Backprop -----
def backward(y, cache, learning_rate=0.1):
    global W1, b1, W2, b2 # Moved global declaration to the beginning
    X, z1, a1, z2, a2 = cache
    m = X.shape[1]

    # خطای خروجی
    dz2 = a2 - y                # (1, m)
    dW2 = (dz2 @ a1.T) / m
    db2 = np.sum(dz2, axis=1, keepdims=True) / m

    # خطای لایه مخفی
    dz1 = (W2.T @ dz2) * sigmoid_derivative(z1)
    dW1 = (dz1 @ X.T) / m
    db1 = np.sum(dz1, axis=1, keepdims=True) / m

    # آپدیت وزن‌ها
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1
    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2

# ----- 5) آزمایش -----
# داده نمونه (XOR)
X = np.array([[0,0,1,1],
              [0,1,0,1]]).T
X = X.T  # (2,4)

y = np.array([[0,1,1,0]])

for epoch in range(2000):
    a2, cache = forward(X)
    backward(y, cache)

print("Predictions after training:", a2)
```
