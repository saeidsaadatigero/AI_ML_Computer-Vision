```python
import numpy as np

x = np.array([1, 2, 3, 4])
y = np.array([2, 4, 6, 8])

def cost_function(w, x, y, lam):
    y_pred = w * x
    mse = np.mean((y_pred - y) ** 2)
    l2_penalty = lam * (w ** 2)
    return mse + l2_penalty
```

```python
def gradient(w, x, y, lam):
    grad_mse = 2 * np.mean((w * x - y) * x)
    grad_l2 = 2 * lam * w
    return grad_mse + grad_l2
```

```python
def train(lam, lr=0.01, epochs=1000):
    w = 0.0
    for _ in range(epochs):
        w -= lr * gradient(w, x, y, lam)
    final_loss = cost_function(w, x, y, lam)
    return w, final_loss
```

```python
for lam in [0, 0.01, 0.1]:
    w, loss = train(lam)
    print(f"lambda={lam} | w={w:.4f} | loss={loss:.4f}")
```
